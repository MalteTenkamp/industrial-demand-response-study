```python
import pandas as pd

df = pd.read_csv(
    "chiron_session6_min230-380_1min.csv",
    header=None,
    names=["timestamp", "index", "value"]
)

df["value"] = pd.to_numeric(df["value"], errors="coerce")

percentiles = [0.55, 0.60, 0.65, 0.70, 0.75, 0.80, 0.85, 0.90, 0.95]

print(f"Median: {df['value'].median():.4f}")

for p in percentiles:
    print(f"{int(p * 100)}%-Perzentil: {df['value'].quantile(p):.4f}")
```
